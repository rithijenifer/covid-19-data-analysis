```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("covid_data.csv")
df.head()
```

```python
df["Date"] = pd.to_datetime(df["Date"])
```

```python
country_cases = df.groupby("Country")["Confirmed"].max()
country_cases
```

```python
india = df[df["Country"] == "India"]
usa = df[df["Country"] == "USA"]
italy = df[df["Country"] == "Italy"]

plt.plot(india["Date"], india["Confirmed"], label="India")
plt.plot(usa["Date"], usa["Confirmed"], label="USA")
plt.plot(italy["Date"], italy["Confirmed"], label="Italy")

plt.xlabel("Date")
plt.ylabel("Confirmed Cases")
plt.title("COVID-19 Trend Analysis")
plt.legend()
plt.show()
```

```python
country_cases.plot(kind="bar")
plt.title("Country-wise COVID-19 Confirmed Cases")
plt.ylabel("Cases")
plt.show()
```
